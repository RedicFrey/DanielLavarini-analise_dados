```python

# Barras: população 2022 por UF
fig, ax = plt.subplots(figsize=(9, 8))
ordem = estado.sort_values("População Censo 2022", ascending=False)
sns.barplot(data=ordem, y="UF", x="População Censo 2022", ax=ax, color="steelblue")
media_populacao = ordem["População Censo 2022"].mean()
ax.axvline(
    x=media_populacao, 
    color="red", 
    linestyle="--", 
    linewidth=2, 
    label=f"Média: {media_populacao/1e6:.2f} M"
)
ax.set_title("População 2022 por UF")
ax.set_xlabel("Habitantes")
ax.set_ylabel("UF")
plt.tight_layout()
plt.show()
```
